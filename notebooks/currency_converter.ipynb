# Currency Converter

## basic code

- Write a program to convert an amount of money from one currency to another using fixed exchange rates.


- The user inputs the amount and selects the currencies for conversion.
- The input should be positive, else errors
- The currency should be also sharing the currency type for conversion

```python
# define conversion rates
exchange_rates = {
    "usd-to-cad": 1.3866, # 1 USD = 1.3866 CAD
    "usd-to-eur": 0.8579, # 1 USD ≈ 0.8579 EUR
    "usd-to-inr": 80, # 1 USD = 80 INR
}

# exchange_rates = {
#     "usd":
#         [{"cad":"1.3866"},
#         {"eur":"0.8579"},
#         {"inr":"80"},]
# }

# exchange_rates["usd"][0]["cad"]
```

```python
def get_currency_details():

    while True:
    # get amount
        try:
            amount = int(input("Enter the amount: "))
            if amount <= 0:
                raise ValueError("Must be positive")
                continue
            break
        except (ValueError, TypeError) as e:
            print(f"Invalid input {e}")

    while True:
    # get currency type
        try:
            source_currency = input("Source currency (USD/EUR/CAD/INR): ").lower()
            if source_currency not in ["usd", "eur", "cad"]:
                raise ValueError("Please select only one of the 'usd', 'eur', 'cad', or 'inr'")
                continue
            break
        except Exception as e:
            print(f"Error: {e} - Please select only one of the 'usd', 'eur', or 'cad'")

    while True:
    # get currency type
        try:
            target_currency = input("Target currency (USD/EUR/CAD): ").lower()
            if target_currency not in ["usd", "eur", "cad"]:
                raise ValueError("Please select only one of the 'usd', 'eur', or 'cad'")
                continue
            if target_currency == source_currency:
                raise ValueError(f"Source={source_currency.upper()}, Target={target_currency.upper()} currencies cannot be same. Try again")
            break
        except Exception as e:
            print(f"Error: {e} - Please select only one of the 'usd', 'eur', or 'cad'")

    # return amount, currency
    return (amount, source_currency, target_currency)

amount, source_currency, target_currency = get_currency_details()
amount, source_currency, target_currency
```

```python
def get_currency_conversion(amount, source, target, exchange_rates):
    try:
        if source == target:
            return amount
        else:
            if source == "usd":
                if target == "cad":
                    convertered_currency = amount * exchange_rates["usd-to-cad"]
                    return convertered_currency
                elif target == "eur":
                    convertered_currency = amount * exchange_rates["usd-to-eur"]
                    return convertered_currency
                elif target == "inr":
                    convertered_currency = amount * exchange_rates["usd-to-inr"]
                    return convertered_currency
                else:
                    raise ValueError()
            elif source == "cad":
                convertered_currency = amount * (1/exchange_rates["usd-to-cad"])
                return convertered_currency
            elif source == "eur":
                convertered_currency = amount * (1/exchange_rates["usd-to-eur"])
                return convertered_currency
            elif source == "inr":
                convertered_currency = amount * (1/exchange_rates["usd-to-inr"])
                return convertered_currency
            else:
                raise ValueError()
    except ValueError:
        print("No conversion rates available for your selection")

get_currency_conversion(100, source="usd", target="cad", exchange_rates=exchange_rates)
```

```python
exchange_rates = {
    "usd": {"cad":"1.3866", "eur":"0.8579"},
    "eur": {"usd":"1.18", "cad":"1.47"},
    "cad": {"usd":"0.80", "eur":"0.68"},
}

def get_currency_details():
    """Get currency conversion details from user input."""

    # Get valid currencies from exchange_rates keys
    VALID_CURRENCIES = list(exchange_rates.keys())

    def get_amount():
        """Prompt user for a valid positive amount."""
        while True:
            try:
                amount = int(input("Enter the amount: "))
                if amount <= 0:
                    raise ValueError("Amount must be positive")
                return amount
            except (ValueError, TypeError) as e:
                print(f"Invalid input: {e}")

    def get_currency(prompt, valid_currencies=VALID_CURRENCIES):
        """Prompt user for a valid currency."""
        while True:
            try:
                currency = input(prompt).lower()
                if currency not in valid_currencies:
                    raise ValueError(f"Please select one of: {', '.join(valid_currencies)}")
                return currency
            except Exception as e:
                print(f"Error: {e}")

    # Get amount
    amount = get_amount()

    # Get source currency
    currency_list = '/'.join([c.upper() for c in VALID_CURRENCIES])
    source_currency = get_currency(f"Source currency ({currency_list}): ")

    # Get target currency (excluding source)
    while True:
        target_currency = get_currency(f"Target currency ({currency_list}): ")
        if target_currency == source_currency:
            print(f"Error: Source and target currencies cannot be the same ({source_currency.upper()}). Try again.")
        else:
            break

    return (amount, source_currency, target_currency)

amount, source_currency, target_currency = get_currency_details()
amount, source_currency, target_currency
```

```python
exchange_rates = {
    "usd": {"cad":"1.3866", "eur":"0.8579"},
    "eur": {"usd":"1.18", "cad":"1.47"},
    "cad": {"usd":"0.80", "eur":"0.68"},
}

def get_currency_conversion(amount, source, target, exchange_rates):
    """Convert amount from source currency to target currency using exchange rates."""
    try:
        # Check if source currency exists in exchange_rates
        if source not in exchange_rates:
            raise ValueError(f"Source currency '{source}' not found in exchange rates")

        # Check if target currency exists for the source
        if target in exchange_rates[source]:
            # Direct conversion available
            rate = float(exchange_rates[source][target])
            converted_currency = amount * rate
            return converted_currency
        else:
            # Try indirect conversion through USD
            if source != "usd" and target != "usd":
                # Convert source to USD first, then USD to target
                if "usd" in exchange_rates[source] and target in exchange_rates["usd"]:
                    # source -> USD
                    usd_amount = amount * float(exchange_rates[source]["usd"])
                    # USD -> target
                    converted_currency = usd_amount * float(exchange_rates["usd"][target])
                    return converted_currency
            raise ValueError(f"No conversion path found from '{source}' to '{target}'")
    except ValueError as e:
        print(f"No conversion rates available for your selection: {e}")
        return None

get_currency_conversion(100, source="usd", target="cad", exchange_rates=exchange_rates)
```
