# Number Guessing Game

## Basic Code

```python
# BASIC

import random


# generate a random number b/w 1,100
generated_number = random.randint(1,100)

# count the number of attempts
count = 0

# run a continuous loop
while True:

    # ask the user to guess a number
    try:
        guessed_number = int(input("Guess the number (between 1 and 100): "))
    except:
        raise ValueError("Please enter a valid Integer")

    # exit the loop if numbers match, else keep asking the user to guess
    if guessed_number == generated_number:
        break
    elif guessed_number > generated_number:
        count += 1
        print(f"{guessed_number}: Too high! Try again.")
    else:
        count += 1
        print(f"{guessed_number}: Too low! Try again.")

print(f"Congratulations! You guessed the correct number {generated_number} in {count} attempts.")
```
