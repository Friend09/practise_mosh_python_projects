## Feature1


Add options for the user to choose the color of the QR code. This will allow users to generate QR codes that match their style or branding.


```python
import qrcode

qr = qrcode.QRCode(
    version=2,
    error_correction=qrcode.ERROR_CORRECT_L,
    box_size=10,
    border=2,
)

qr.add_data('Some data')
qr.make(fit=True)

img = qr.make_image(back_color="gray", fill_color="blue")
img
```

```python
def get_qr_details():
    image_data = input("Enter the image data for the QR Code: ").strip()
    image_name = input("Enter the image name: ").strip()
    back_color = input("Pick a background color for QR code: ['white', 'gray', 'blue']").lower()
    fill_color = input("Pick a color for QR code pixels: ['white', 'gray', 'blue']").lower()
    return (image_data, image_name, back_color, fill_color)

image_data, image_name, back_color, fill_color = get_qr_details()
print(image_data, img_name, back_color, fill_color)
```

```python
def generate_qrcode(
        data: str,
        img_name: str,
        back_color: str,
        fill_color: str,
        file_path: str = "/Users/vamsi_mbmax/Developer/VAM_Documents/01_vam_PROJECTS/LEARNING/proj_PersonalProjects/dev/pract_pp_mosh_python_projects/files"
    ) -> str:

    import qrcode
    qrcode.QRCode(
        version=1,
        error_correction=qrcode.ERROR_CORRECT_L,
        box_size=5,
        border=2,
    )
    qr.add_data(data)
    qr.make(fit=True)

    img = qr.make_image(fill_color=fill_color, back_color=back_color)

    img.save(file_path+"/"+img_name)
    return f"QR code saved as {file_path+"/"+img_name}"

generate_qrcode(
    data=image_data,
    img_name=image_name,
    back_color=back_color,
    fill_color=fill_color,
)
```
