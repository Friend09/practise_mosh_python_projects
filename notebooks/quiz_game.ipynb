# QUIZ GAME

## Basic Game

- Write a program to create a quiz where the player answers multiple-choice questions. 
- The program should evaluate the player's answers and provide a score at the end.

- discussion points
  - the user will be shown with a first question w/ options
    - will always be the same question
  - 3 questions will be there inside a questions dictionary
    - each question will have 3 options
    - only one option shall be selected
    - for correct answer, show "Correct!"
    - for incorrect answer show "incorrect!"
    - loop in the current question until a valid response is provided
  - once all questions are complete, show "Your final score: /total_questions"

- Approach
  - a dictionary to hold a set of questions, answers
  - a counter to keep track on the score
  - run a loop over the questions dict
    - present the user with the first question, options
    - user to respond
    - response will be checked for validity - loop until valid response
    - compare w/ the answer, show the result and go to the next question
    - once all questions are complete, exit the loop
  - show the final score

```python
while counter < total_questions:
    counter += 1
    # print(questions[counter])
    user_response = input(f"Question {counter}: {questions[counter]['question']}\n {questions[counter]['options']}:\nEnter answers in 'a, b, c, ..'")

    if user_response == questions[counter]['answer']:
        score += 1
        result = colored("Correct!", "green")
        print(f"Your answer: {user_response}")
        print(result)
    else:
        result = colored("Incorrect", "red")
        print(f"Your answer: {user_response}")
        print(result)

print(f"\nYour final scores: {score}/{total_questions}")
```

## improve basic code

```python
from typing import Dict, List
from pydantic import BaseModel

class Question(BaseModel):
    question: str
    options: List[str]
    answer: str

def get_questions() -> Dict[int, Question]:
    questions = {
        1: Question(
            question="What is the capital of France?",
            options=["Paris", "London", "Rome"],
            answer="a"
        ),
        2: Question(
            question="What is the largest planet in our solar system?",
            options=["Earth", "Jupiter", "Mars"],
            answer="b"
        ),
        3: Question(
            question="What is the smallest prime number?",
            options=["0", "1", "2"],
            answer="c"
        ),
        4: Question(
            question="Which ocean is the largest?",
            options=["Atlantic", "Indian", "Pacific"],
            answer="c"
        ),
    }
    return questions

get_questions()
```

```python
def get_user_responses(
    total_questions: int = len(questions),
    score: int = 0,
    counter: int = 0,
    ) -> (int, int):

    while counter < total_questions:
        counter += 1
        # print(questions[counter])
        user_response = input(f"Question {counter}: {questions[counter]['question']}\n {questions[counter]['options']}:\nEnter answers in 'a, b, c, ..'")

        if user_response == questions[counter]['answer']:
            score += 1
            result = colored("Correct!", "green")
            print(f"Your answer: {user_response}")
            print(result)
        else:
            result = colored("Incorrect", "red")
            print(f"Your answer: {user_response}")
            print(result)

    return (score,total_questions)

result = get_user_responses()
print(f"\nYour final scores  (correct/total): {result[0]}/{result[1]}")
```
