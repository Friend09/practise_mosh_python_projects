## feature1: first player (or computer) to win

```python
## feature1: first player (or computer) to win two out of three rounds is declared the overall winner.

import random

# count the total games (skip the TIE)
total_games = 0
# count the player wins
user_wins = 0
computer_wins = 0

# prompt the user to select
while True:
    user_choice = input("Rock 🪨, paper 📃, Or scissors ✂? choose (r/p/s): ")
    user_choice = user_choice.lower()
    if user_choice not in ["r", "p", "s"]:
        print("please choose between (r/p/s)")
        continue

    # generate computer choice
    computer_choice = random.choice(["r", "p", "s"])

    # chose winner (game logic)
    if (user_choice == "r" and computer_choice == "s"):
        print(f"You Chose 🪨 \n Computer chose ✂")
        print(f"You Win!")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == "p" and computer_choice == "r"):
        print(f"You Chose 📃 \n Computer chose 🪨")
        print(f"You Win!")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == "s" and computer_choice == "p"):
        print(f"You Chose ✂ \n Computer chose 📃")
        print(f"You Win!")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == computer_choice):
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"It’s a TIE")
    else:
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"You lose")
        total_games += 1
        computer_wins += 1
        print(f"total games played so far are : {total_games}\n")

    # declare final winner
    if total_games == 3:
        if user_wins > computer_wins:
            print(f"Winner is USER with: {user_wins}")
            break
        elif computer_wins > user_wins:
            print(f"Winner is COMPUTER with: {computer_wins} Wins!!")
            break

    # prompt the user to continue?
    while True:
        to_continue = input("Continue? (y/n): ")
        to_continue = to_continue.lower()
        if to_continue in ["y", "n"]:
            break
        print("please enter y/n for yes or no to continue")
    if to_continue ==  "n":
        break
```

```python
## feature2: Keep a tally of how many times the player wins, loses, or ties with the computer

import random

# count the total games (skip the TIE)
total_games = 0
# count the player wins
user_wins = 0
computer_wins = 0
ties_w_computer = 0

# prompt the user to select
while True:
    user_choice = input("Rock 🪨, paper 📃, Or scissors ✂? choose (r/p/s): ")
    user_choice = user_choice.lower()
    if user_choice not in ["r", "p", "s"]:
        print("please choose between (r/p/s)")
        continue

    # generate computer choice
    computer_choice = random.choice(["r", "p", "s"])

    # chose winner (game logic)
    if (user_choice == "r" and computer_choice == "s"):
        print(f"You Chose 🪨 \n Computer chose ✂")
        print(f"You Win!\n")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == "p" and computer_choice == "r"):
        print(f"You Chose 📃 \n Computer chose 🪨")
        print(f"You Win!\n")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == "s" and computer_choice == "p"):
        print(f"You Chose ✂ \n Computer chose 📃")
        print(f"You Win!\n")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == computer_choice):
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"It’s a TIE")
        total_games += 1
        ties_w_computer += 1
        print(f"total games played so far are : {total_games}\n")
    else:
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"You lose")
        total_games += 1
        computer_wins += 1
        print(f"total games played so far are : {total_games}\n")

    # declare final winner
    if user_wins >= 2 or computer_wins >= 2:
        if user_wins > computer_wins:
            print(f"Winner is USER with: {user_wins}")
            print(f"Total ties: {ties_w_computer}")
            print(f"total games played so far: {total_games}")
            break
        elif computer_wins > user_wins:
            print(f"Winner is COMPUTER with: {computer_wins} Wins!!")
            print(f"Total ties: {ties_w_computer}")
            print(f"total games played so far: {total_games}")
            break

    # prompt the user to continue?
    while True:
        to_continue = input("Continue? (y/n): ")
        to_continue = to_continue.lower()
        if to_continue in ["y", "n"]:
            break
        print("please enter y/n for yes or no to continue")
    if to_continue ==  "n":
        break
```

## feature3: two players


```python
## feature3: Add an option for two players to play against each other, taking turns to input their choices

import random

# Constants
CHOICE_EMOJI = {"r": "🪨", "p": "📃", "s": "✂"}
VALID_CHOICES = ["r", "p", "s"]
WINNING_COMBINATIONS = {
    ("r", "s"): True,
    ("p", "r"): True,
    ("s", "p"): True
}

def get_user_choice(user_name):
    """Get and validate user choice for rock, paper, or scissors."""
    while True:
        choice = input(f"Welcome {user_name}. Please select:\n Rock 🪨, paper 📃, Or scissors ✂? choose (r/p/s): ").lower()
        if choice in VALID_CHOICES:
            return choice
        print(f"Hey {user_name}: please choose between (r/p/s)")

def get_choice_emoji(choice):
    """Convert choice letter to emoji."""
    return CHOICE_EMOJI.get(choice, choice)

def determine_winner(user_1_choice, user_2_choice):
    """Determine the winner between two players.
    Returns: 'user_1', 'user_2', or 'tie'
    """
    if user_1_choice == user_2_choice:
        return 'tie'

    if WINNING_COMBINATIONS.get((user_1_choice, user_2_choice)):
        return 'user_1'
    return 'user_2'

def display_round_result(user_1_name, user_1_choice, user_2_name, user_2_choice, winner, total_games):
    """Display the result of a single round."""
    user_1_emoji = get_choice_emoji(user_1_choice)
    user_2_emoji = get_choice_emoji(user_2_choice)

    print(f"{user_1_name} Chose {user_1_emoji} \n {user_2_name} chose {user_2_emoji}")

    if winner == 'tie':
        print("It's a TIE")
    elif winner == 'user_1':
        print(f"{user_1_name} Wins!\n")
    else:
        print(f"{user_2_name} Wins!\n")

    print(f"total games played so far are : {total_games}\n")

def display_final_results(user_1_name, user_1_wins, user_2_name, user_2_wins, ties, total_games):
    """Display the final game results."""
    if user_1_wins > user_2_wins:
        print(f"Winner is {user_1_name} with: {user_1_wins} wins!")
    else:
        print(f"Winner is {user_2_name} with: {user_2_wins} wins!")
    print(f"Total ties: {ties}")
    print(f"total games played so far: {total_games}")

def ask_to_continue():
    """Ask user if they want to continue playing."""
    while True:
        to_continue = input("Continue? (y/n): ").lower()
        if to_continue in ["y", "n"]:
            return to_continue == "y"
        print("please enter y/n for yes or no to continue")

def play_two_player_game(user_1_name="Player 1", user_2_name="Player 2", wins_needed=2):
    """Main game loop for two-player mode."""
    total_games = 0
    user_1_wins = 0
    user_2_wins = 0
    ties = 0

    while True:
        # Get both players' choices
        user_1_choice = get_user_choice(user_1_name)
        user_2_choice = get_user_choice(user_2_name)

        # Determine winner and update scores
        winner = determine_winner(user_1_choice, user_2_choice)
        total_games += 1

        if winner == 'user_1':
            user_1_wins += 1
        elif winner == 'user_2':
            user_2_wins += 1
        else:
            ties += 1

        # Display round result
        display_round_result(user_1_name, user_1_choice, user_2_name, user_2_choice, winner, total_games)

        # Check if someone won the match
        if user_1_wins >= wins_needed or user_2_wins >= wins_needed:
            display_final_results(user_1_name, user_1_wins, user_2_name, user_2_wins, ties, total_games)
            break

        # Ask to continue
        if not ask_to_continue():
            break

# Run the game
play_two_player_game()
```
