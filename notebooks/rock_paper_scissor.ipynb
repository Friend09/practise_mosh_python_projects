## Basic

```python
import random

# prompt the user to select
while True:
    user_choice = input("Rock 🪨, paper 📃, Or scissors ✂? choose (r/p/s): ")
    user_choice = user_choice.lower()
    if user_choice not in ["r", "p", "s"]:
        print("please choose between (r/p/s)")
        continue

    # generate computer choice
    computer_choice = random.choice(["r", "p", "s"])

    # chose winner (game logic)
    if (user_choice == "r" and computer_choice == "s"):
        print(f"You Chose 🪨 \n Computer chose ✂")
        print(f"You Win!")
    elif (user_choice == "p" and computer_choice == "r"):
        print(f"You Chose 📃 \n Computer chose 🪨")
        print(f"You Win!")
    elif (user_choice == "s" and computer_choice == "p"):
        print(f"You Chose ✂ \n Computer chose 📃")
        print(f"You Win!")
    elif (user_choice == computer_choice):
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"It’s a TIE")
    else:
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"You lose")

    # prompt the user to continue?
    while True:
        to_continue = input("Continue? (y/n): ")
        to_continue = to_continue.lower()
        if to_continue in ["y", "n"]:
            break
        print("please enter y/n for yes or no to continue")
    if to_continue ==  "n":
        break
```

## feature1: first player (or computer) to win

```python
## feature1: first player (or computer) to win two out of three rounds is declared the overall winner.

import random

# count the total games (skip the TIE)
total_games = 0
# count the player wins
user_wins = 0
computer_wins = 0

# prompt the user to select
while True:
    user_choice = input("Rock 🪨, paper 📃, Or scissors ✂? choose (r/p/s): ")
    user_choice = user_choice.lower()
    if user_choice not in ["r", "p", "s"]:
        print("please choose between (r/p/s)")
        continue

    # generate computer choice
    computer_choice = random.choice(["r", "p", "s"])

    # chose winner (game logic)
    if (user_choice == "r" and computer_choice == "s"):
        print(f"You Chose 🪨 \n Computer chose ✂")
        print(f"You Win!")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == "p" and computer_choice == "r"):
        print(f"You Chose 📃 \n Computer chose 🪨")
        print(f"You Win!")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == "s" and computer_choice == "p"):
        print(f"You Chose ✂ \n Computer chose 📃")
        print(f"You Win!")
        total_games += 1
        user_wins += 1
        print(f"total games played so far are : {total_games}\n")
    elif (user_choice == computer_choice):
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"It’s a TIE")
    else:
        print(f"You Chose {user_choice} \n Computer chose {computer_choice}")
        print(f"You lose")
        total_games += 1
        computer_wins += 1
        print(f"total games played so far are : {total_games}\n")

    # declare final winner
    if total_games == 3:
        if user_wins > computer_wins:
            print(f"Winner is USER with: {user_wins}")
            break
        elif computer_wins > user_wins:
            print(f"Winner is COMPUTER with: {computer_wins} Wins!!")
            break

    # prompt the user to continue?
    while True:
        to_continue = input("Continue? (y/n): ")
        to_continue = to_continue.lower()
        if to_continue in ["y", "n"]:
            break
        print("please enter y/n for yes or no to continue")
    if to_continue ==  "n":
        break
```
