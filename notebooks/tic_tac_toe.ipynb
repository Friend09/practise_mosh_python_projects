# [8. Tic Tac Toe Game]


- Write a program to create a classic Tic Tac Toe game.
- Two players take turns marking Xs and Os on a 3x3 grid
- The first player to align three of their marks either vertically, horizontally, or diagonally wins the game.
- The board is printed after each move, with colored marks to distinguish between the players.
- The game checks for a winner after each move and ends when a player wins or when the board is full, resulting in a draw.


## basic code


```python
# == get row, col


def get_row_col_values():
    # get row
    while True:
        row = int(input("Enter row (0-2): "))
        if row not in [0, 1, 2]:
            print(f"please enter a number (0-2): {row}")
            continue
        break

    # get col
    while True:
        col = int(input("Enter col (0-2): "))
        if col not in [0, 1, 2]:
            print(f"please enter a number (0-2): {col}")
            continue
        break
    return (row, col)


row, col = get_row_col_values()
print(row, col)
```

```python
# == get player
player_index = 0
players = ["X", "O"]

# to toggle a player
# player_index = not player_index


# get row, col, player
print(players[player_index])
```

```python
# check if board cell is empty

board_matrix = [
    ["", "", ""],
    ["", "", ""],
    ["", "", ""],
]

# add to board
# make sure row/col exist before using them
try:
    _ = board_matrix[row][col]
except (NameError, IndexError, TypeError):
    row, col = get_row_col_values()

while board_matrix[row][col] != "":
    print("The board cell is not empty, please pick a new cell")
    row, col = get_row_col_values()

board_matrix[row][col] = players[player_index]
player_index = int(not player_index)
print(board_matrix)
```
