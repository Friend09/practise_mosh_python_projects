# TO-DO List

> Write a program to create a simple To-Do List application. 

- The program allows users to add tasks
- view the current list of tasks
- and remove tasks once they are completed.

```yaml
Todo List Menu:
1. View Tasks
2. Add a Task
3. Remove a Task
4. Exit
Enter your choice: 2
Enter a new task: Buy groceries
```

## basic code

````python
# - user tasks list
task_list = []
````

````python
# - function - add a task
def add_a_task(user_input: str, task_list) -> None:
    task_list.append(user_input)
    print(f"task added. here is your updated task list: {task_list}")
````

````python
# - function - remove a task
def remove_a_task(task_list) -> None:
    print(task_list)
    while True:
        if len(task_list) == 0:
            print(f"Task list is empty: {len(task_list)}")
            break
        try:
            user_item = int(input("Enter your Task"))
            for idx, item in enumerate(task_list):
                # print(idx, item)
                if user_item == idx + 1:
                    removed_item = task_list.pop(idx)
                    print(f"removed task: {removed_item}")
                    break
        except ValueError:
            print("Please enter a valid task number")
````

````python
task_list
````

````python
# - loop for user input
while True:
    #     - get user input
    user_action = int(input("""
        Todo List Menu:
            1. View Tasks
            2. Add a Task
            3. Remove a Task
            4. Exit
        Enter your choice:
    """))

    # view tasks
    if user_action == 1:
        print(task_list)
    # add a task
    elif user_action == 2:
        task = input("Add your task: ")
        add_a_task(task, task_list)
    # remove a task
    elif user_action == 3:
        remove_a_task(task_list)
    # - respond / exit
    elif user_action == 4:
        break
````
